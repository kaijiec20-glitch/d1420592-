```python
# 固定的兩個活動名稱，以 tuple 保存。
activity_names = ("桌遊體驗", "電影欣賞")

# 收到的原始文字名單，請保留原值。
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 用活動名稱查詢每人費用。
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}
```

```python
# 1. 處理「桌遊體驗」名單
# 移除空白並轉為小寫
cleaned_board = board_raw.replace(" ", "").lower()
# 以逗號分割並使用 set 去重
board_set = set(cleaned_board.split(","))
# 排序名單
sorted_board = sorted(list(board_set))

# 2. 處理「電影欣賞」名單
cleaned_movie = movie_raw.replace(" ", "").lower()
movie_set = set(cleaned_movie.split(","))
sorted_movie = sorted(list(movie_set))

# 3. 計算各項指標
# 各場活動人數
num_board = len(board_set)
num_movie = len(movie_set)

# 總費用
fee_board = num_board * unit_fees["桌遊體驗"]
fee_movie = num_movie * unit_fees["電影欣賞"]
total_fees = fee_board + fee_movie

# 報名人次 (兩場活動各自去重後的人數相加)
total_registrations = num_board + num_movie

# 總不重複人數 (聯集)
unique_participants = len(board_set.union(movie_set))

# 4. 顯示結果
print(f"【{activity_names[0]}】")
print(f"整理後名單：{sorted_board}")
print(f"參加人數：{num_board} 人，費用小計：{fee_board} 元\n")

print(f"【{activity_names[1]}】")
print(f"整理後名單：{sorted_movie}")
print(f"參加人數：{num_movie} 人，費用小計：{fee_movie} 元\n")

print("【總計數據】")
print(f"總不重複人數：{unique_participants} 人")
print(f"報名總人次：{total_registrations} 人次")
print(f"應收總費用：{total_fees} 元")
```

```python
# 1. 讀取活動名稱
activity_1, activity_2 = activity_names

# 2. 整理名單：轉小寫、移除半形空白，並依逗號拆分（保留重複與原始順序）
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 3. 顯示原始整理後的名單與活動名稱
print(f"活動 1：{activity_1}")
print(f"原始報名名單（已轉小寫、去空白）：{board_list}\n")

print(f"活動 2：{activity_2}")
print(f"原始報名名單（已轉小寫、去空白）：{movie_list}")
```

```python
# 1. 將串列轉為集合 (自動去重)
set_board = set(board_list)
set_movie = set(movie_list)

# 2. 進行集合運算並排序結果
# 共同參加者 (交集)
common_participants = sorted(list(set_board.intersection(set_movie)))

# 全部參加者 (聯集)
all_participants = sorted(list(set_board.union(set_movie)))

# 只參加桌遊者 (桌遊差集電影)
only_board = sorted(list(set_board.difference(set_movie)))

# 只參加電影者 (電影差集桌遊)
only_movie = sorted(list(set_movie.difference(set_board)))

# 3. 顯示運算結果
print("【集合運算與名單比較結果】")
print(f"共同參加者 (交集)：{common_participants}")
print(f"全部不重複參加者 (聯集)：{all_participants}")
print(f"只參加桌遊者 (差集 - 桌遊但無電影)：{only_board}")
print(f"只參加電影者 (差集 - 電影但無桌遊)：{only_movie}")
```

```python
# 1. 取得排序後的全部參加者（前面已排序過，此處再次確認排序以確保完整性）
sorted_all_participants = sorted(list(set_board.union(set_movie)))

# 2. 使用切片取得前兩位成員
preview_list = sorted_all_participants[:2]

# 3. 顯示結果
print("【通知名單預覽】")
print(f"全部參加者名單（排序後）：{sorted_all_participants}")
print(f"前兩位預覽（切片 [:2]）：{preview_list}")
```

```python
# 1. 建立 summary 字典儲存所有指標數據
summary = {
    "桌遊人數": len(set_board),
    "電影人數": len(set_movie),
    "共同人數": len(set_board.intersection(set_movie)),
    "總不重複人數": len(set_board.union(set_movie)),
    "報名人次": len(set_board) + len(set_movie),
    "桌遊應收": len(set_board) * unit_fees["桌遊體驗"],
    "電影應收": len(set_movie) * unit_fees["電影欣賞"],
}

# 計算總應收費用並加入字典
summary["總應收"] = summary["桌遊應收"] + summary["電影應收"]

# 2. 從 summary 讀取數值並格式化輸出摘要
print("【活動報名摘要報告】")
print(f"桌遊體驗參加人數：{summary['桌遊人數']} 人")
print(f"電影欣賞參加人數：{summary['電影人數']} 人")
print(f"共同參加兩場人數：{summary['共同人數']} 人")
print(f"總計不重複人數：{summary['總不重複人數']} 人")
print(f"總計報名人次：{summary['報名人次']} 人次")
print("-" * 30)
print(f"桌遊體驗應收費用：NT$ {summary['桌遊應收']:,} 元")
print(f"電影欣賞應收費用：NT$ {summary['電影應收']:,} 元")
print(f"總計應收費用：NT$ {summary['總應收']:,} 元")
```

```python
# 計算精確到小數點後兩位的各項金額
board_total_fee = len(set_board) * unit_fees["桌遊體驗"]
movie_total_fee = len(set_movie) * unit_fees["電影欣賞"]
overall_total_fee = board_total_fee + movie_total_fee

print("==================== 活動報名完整摘要報告 ====================")
# 1. 活動名稱及每人費用
print(f"活動 1：{activity_names[0]} (每人費用: NT$ {unit_fees['桌遊體驗']:.2f} 元)")
print(f"活動 2：{activity_names[1]} (每人費用: NT$ {unit_fees['電影欣賞']:.2f} 元)")
print("-" * 60)

# 2. 兩份整理後、仍保留重複項目的串列
print(f"桌遊體驗原始名單 (保留重複)：{board_list}")
print(f"電影欣賞原始名單 (保留重複)：{movie_list}")
print("-" * 60)

# 3. 兩場各自去重並排序的名單與人數
print(f"桌遊體驗去重排序名單：{sorted_board} (共 {len(set_board)} 人)")
print(f"電影欣賞去重排序名單：{sorted_movie} (共 {len(set_movie)} 人)")
print("-" * 60)

# 4. 共同參加者、只參加桌遊者、只參加電影者
print(f"共同參加兩場者 (交集)：{common_participants if common_participants else []}")
print(f"只參加桌遊體驗者 (差集)：{only_board if only_board else []}")
print(f"只參加電影欣賞者 (差集)：{only_movie if only_movie else []}")
print("-" * 60)

# 5. 全部參加者、總不重複人數、報名人次及前兩位通知名單
print(f"全部不重複參加者 (聯集)：{sorted_all_participants}")
print(f"總不重複人數：{len(sorted_all_participants)} 人")
print(f"總報名人次：{len(board_list) + len(movie_list)} 人次")
print(f"前兩位通知名單 (切片預覽)：{preview_list}")
print("-" * 60)

# 6. 各場應收與總應收，金額顯示至小數點後兩位
print(f"桌遊體驗總應收：NT$ {board_total_fee:.2f} 元")
print(f"電影欣賞總應收：NT$ {movie_total_fee:.2f} 元")
print(f"所有活動總應收：NT$ {overall_total_fee:.2f} 元")
print("==============================================================")
```

```python
def run_test_suite(test_id, test_board_raw, test_movie_raw, expected_total_fee, expected_unique):
    # 1. 整理名單
    t_board_list = test_board_raw.replace(" ", "").lower().split(",")
    t_movie_list = test_movie_raw.replace(" ", "").lower().split(",")

    t_set_board = set(t_board_list)
    t_set_movie = set(t_movie_list)

    # 2. 計算指標
    t_board_fee = len(t_set_board) * unit_fees["桌遊體驗"]
    t_movie_fee = len(t_set_movie) * unit_fees["電影欣賞"]
    t_total_fee = t_board_fee + t_movie_fee

    t_unique = len(t_set_board.union(t_set_movie))

    # 3. 驗證與顯示結果
    print(f"=== 測試組 {test_id} ===")
    print(f"桌遊原始輸入: {test_board_raw.strip()}")
    print(f"電影原始輸入: {test_movie_raw.strip()}")
    print(f"計算總費用: NT$ {t_total_fee} (期望值: {expected_total_fee})")
    print(f"不重複人數: {t_unique} (期望值: {expected_unique})")

    assert t_total_fee == expected_total_fee, f"測試組 {test_id} 費用不符！"
    assert t_unique == expected_unique, f"測試組 {test_id} 不重複人數不符！"
    print("=> 測試通過！\n")

# 測試資料一：全新名單，完全不重複
run_test_suite(
    test_id=1,
    test_board_raw=" Alice, Bob ",
    test_movie_raw=" Charlie, David ",
    expected_total_fee=2 * 120 + 2 * 80,  # 240 + 160 = 400
    expected_unique=4
)

# 測試資料二：單一活動有多個重複名稱，且兩邊皆有共同參加者
run_test_suite(
    test_id=2,
    test_board_raw=" Fiona, Fiona, Fiona, Gaby ",
    test_movie_raw=" Gaby, Helen, Helen ",
    expected_total_fee=2 * 120 + 2 * 80,  # 240 + 160 = 400
    expected_unique=3  # Fiona, Gaby, Helen
)

# 測試資料三：有一邊活動完全無人報名 (空字串經整理後長度為1，故測試單人與全空名單的組合)
run_test_suite(
    test_id=3,
    test_board_raw=" Ivan ",
    test_movie_raw=" Ivan ",
    expected_total_fee=1 * 120 + 1 * 80,  # 120 + 80 = 200
    expected_unique=1
)
```
